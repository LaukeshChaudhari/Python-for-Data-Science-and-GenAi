#### Pandas-DataFrame And Series

Pandas is a known data manipulation library in Python, widely used for data analysis and data cleanong. it provides two primary data structure: Series and DataFrame. a series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
# Series
# A Pandas Series is a one-dimensional array-like object that can hold any data type. it is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
```

```python
# Create a Series from dictionary
data={'a':1,'b':2, 'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
# DataFrame
# Create a Dataframe from a dictionary of list

data={
    'Name':['Love','John','Jack'],
    'Age':[25,30,35],
    'City':['Banglore','New York','Florida']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
# Create a Data frame From a List of Dictionaries

data=[
    {'Name':'Love','Age':25,'City':'Banglore'},
    {'Name':'John','Age':35,'City':'New York'},
    {'Name':'Jacob','Age':45,'City':'Florida'},
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
