```python
# WAF to print the length of a list. (list is the paramemter)

# WAF to print the elements of a list in asingle line. (list is the paramemter)

cities = ["delhi", "gurgaon", "noida","pune","mumbai"]
heroes = ["thor","ironman","captain","shaktiman"]

def print_len(list):
    print(len(list))

def print_list(list):
    for item in list:
        print(item, end=" ")

print_list(heroes)
print_len(cities)
# print()
```

```python
# WAF to find the factorial of n. (n is the parameter)

def cal_fact(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    print(fact)

cal_fact(6)
```

```python
# Keywords Arguments

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Postional Arguments : {val}")

    for key,value in kwargs.items():
        print(f"Keywords Arguments : {key}:{value}")
```

```python
print_details(1,2,3,4,"Love",name="Love",age="23",country="India")
```

```python
# Return statements
def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
# Recursion 
# prints n to 1 backwords

def show(n):
    if(n == 0): # Base Case
        return
    print(n)
    show(n-1)
    print("End")

show(5)
```

```python
# Factorial

def fact(n):
    if(n == 0 or n == 1):
        return 1
    return n * fact(n-1)
    
fact(4)
```

```python
# What a recursion function to calculate the sum of first n natural numbers.

def natural_num(n):
    if(n == 0):
        return 0
    return natural_num(n - 1) + n

sum = natural_num(5)
print(sum)
```

```python
# num / 2 + num 

def even_num(n):
    if(n%2==0):
        return 0
    return even_num(n - 1)

sum = even_num(20)
print(sum)
```

```python
# write a recursive fun to print all elements in a list.
def print_list(list, idx=0):
    if(idx == len(list)):
        return
    print(list[idx])
    print_list(list, idx+1)

fruits = ["mango", "litchi","apple","banana"]
print_list(fruits)
```
