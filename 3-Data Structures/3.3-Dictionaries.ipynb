### Dictionaries

#### Introduction to Dictionaries

- Distionaries are unordered collection of items. They store data in kay-value pairs. Keys must be unique and immutable (eg. strings, numbers, or tuples), while values can be of any type.

```python
# Creating Dictionaris 

empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Love","age":23,"grade":24}
print(student)
print(type(student))
```

```python
info = {
    "name" : "Love",
    "subjects" : ["python", "c", "java"],
    "topics" : ("dict", "set"),
     "age" : 35,
     "is_adult" : True,
     12.99 : 94.4
}

print(info)
```

```python
# Single Key is Always Used 
student={"name":"Love","age":23,"name":24}
print(student)
```

```python
# accesing Dictionary elements

student={"name":"Love","age":23,"grade":'A'}
print(student)
```

```python
# Accessing Dictionary Elements

print(student['grade'])
print(student['age'])

# Accesing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name,"Not Available'))
```

```python
# Modifying Dictionary Elements
# Dictionary are mutable, so you can add, update or delete elements
print(student)
```

```python
student["age"]=22 #update value for the key 
print(student)
student["address"]="India" # Added a new key and value
print(student)
```

```python
del student['grade'] #delete key and value pair
print(student)
```

```python
# Dictionary Methods

keys=student.keys() #get all the keys
print(keys)
Values=student.values() #get all key value
print(Values)

items=student.items() # get all key value pairs
print(items)

items=student.get('grade') # return the key according to value 
print(items)

items = {"city" : "pune"}
student.update(items)  # inserts the specified items to the dictionaries
print(items)
```

```python
# shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student_copy["name"]="Lov3"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() # Shallow Copy
print(student_copy1)
print(student)
```

```python
student["name"]="Lov3"
print(student_copy1)
print(student)
```

```python
# Iterating Over Dictionaries
# YOu can use loops to iterate over dictionaries, key, value, or items

# Iterating over.keys
for keys in student.keys():
    print(keys)
```

```python
# Iterate over values 
for value in student.values():
    print(value)
```

```python
# Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
# Nested Dictionaries
students={
    "student1":{"Name":"Love","age":23},
    "student2":{"name":"Tushar","age":28}
}
print(students)
```

```python
# Access nested dictionaries elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
students.items()
```

```python
# Iterating Over nested Dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
# Dictionary Comprehension
squares={x:x**2 for x in range (5)}
print(squares)
```

```python
# Condition dictionary comprehension
evens={x:x**2 for x in range (10) if x%2==0}
print(evens)
```

```python
# Practicle Examples

#use a dictionary to count the frequency of elements in lis

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
# Merge 2 Dictionaries into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
