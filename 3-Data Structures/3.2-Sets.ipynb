### Sets

Sets are built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a sdpecific order, and they do not allow duplicate element. Sets are useful for memebership tests, eliminating duplicate entries, anmd performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
# create a set 

my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
collection = {1,2,3,4, "hello","world","world",4,2,2,}
print(collection)
print(len(collection))
```

```python
my_empty_set=set() # empty set ; syntax
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
# Basic Set Operation

# Adding And Removing Elements

my_set.add(7)
print(my_set)
my_set.remove(5)
print(my_set)
```

```python
# Remove The Element from a Set 
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(11)
print(my_set)
```

```python
# Pop Method 
removed_elements=my_set.pop()
print(removed_elements)
print(my_set)
```

```python
# Clear All The Elements 
my_set.clear()
print(my_set)
```

```python
# Set Membership test 
my_set.clear()
print(my_set)
```

```python
# Set Membership Test

my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
# Mathematical Operatiopn 

set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# Union
union_set=set1.union(set2)
print(union_set)

# Intersection
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

# Difference 
print(set1.difference(set2))
```

```python
# Symmetric Difference 

set1={1,2,3,4,5}
set2={3,4,5}

# is subset
print(set1.issubset(set2))

print(set1.issuperset(set2))
```

```python
# Conting Unique Word in Text

text="In this Tutorial we are discussing about sets"

words=text.split()

# Convert list if words to set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
