```python
# Loop Control statement 

# break 

# The braek statement exists the loop prematurely

# break statement 

for i in range (10):
    if i==5:
        break
    print(i)
```

```python
num = (1,4,9,16,25,36,49,64,81,100)

x = 36

i = 0
while i < len(num):
    if(num[i] == x):
        print("found at index i : ",i)
        break
    else:
        print("finding.....")
    i += 1
```

```python
# Continue 

# th econtinue statement skips the current iteration and contineous with the next.

for i in range(10):
    if i%2==0:
        continue #skip
    print(i)
```

```python
i = 0
while i <= 5:
    if(i%2 != 0):
        i += 1
        continue
    print(i)
    i += 1
```

```python
# Pass 

# The Pass statement is null operation; it does nothing.

for i in range (5):
    if i==3:
        pass
    print(i)
```

```python
for i in range (5):
    pass
print("some useful work")
```

```python
# Nested Loops 

# a loop inside a loop 

for  i in range(5):
    for j in range(3):
        print(f"i:{i} and j:{j}")
```

```python
# Examples - Calculate the sum of first N natural numbers using a while and for loop

# While Loop

n=10
sum=0
count=1

while count<=n:
    sum+=count
    count+=1
print("Sum of first 10 natural numbers: ",sum)
```

```python
# For Loop

n=10
sum=0
for i in range (11):
    sum=sum+i
print(sum)
```

```python
# Example - Prime Number Between 1 and 100

for num in range (1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
